```python
matrix = [
    [2, 11, 7, 12],
    [5, 2, 9, 15],
    [8, 3, 10, 42]
]

for i in range(len(matrix)):
    row_sum = 0

    for j in range(len(matrix[i])):
        row_sum += matrix[i][j]

    print("Sum of row", i + 1, "=", row_sum)
```
